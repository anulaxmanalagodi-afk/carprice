```python
print(df.isnull())
```

```python
# identify  missing values
print(df.isnull().sum())
```

```python
# all missing values count
print(df.isnull().sum().sum())
```

```python
# that only one column missing using values count
df["Embarked"].isnull().sum()            
```

```python
# find the most common values
df["Embarked"].mode()[0]
```

```python
print(df.isnull().sum())
```

```python
# fill the missing values using  most common value 
df["Embarked"]=df["Embarked"].fillna(df["Embarked"].mode()[0])
```

```python
df["Embarked"].isnull().sum()            
```

```python
print(df.isnull().sum())
```

```python
# fill the missing values using average
df["Age"]=df["Age"].fillna(df["Age"].mean())
```

```python
print(df.isnull().sum())
```

```python
# replace missing values in fillna method
df["Cabin"]=df["Cabin"].fillna("unkown")
```
