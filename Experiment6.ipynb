```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("C:/Users/gptcs/anu/titan.csv")
```

```python
df
```

```python
# method 1:using IQR 
Q1=df["Age"].quantile(0.25)
Q2=df["Age"].quantile(0.75)
```

```python
IQR=Q2-Q1
print("IQR=",IQR)
```

```python
# the Identify the outliers using scatter plot
plt.scatter(df["Age"],df["Fare"],color="blue")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()
```

```python
# the Idenfy outliers using boxplot
plt.boxplot(df["Fare"].dropna())
plt.title("Fare outliears")
plt.ylabel("Fare")
plt.show()
```

```python
# the Identify the outliers in IQR method 
Q1=df["Fare"].quantile(0.25)
Q2=df["Fare"].quantile(0.75)
```

```python
IQR=Q2-Q1
```

```python
lower=Q1 - 1.5*IQR
upper=Q2 - 1.5*IQR
outliers=df[(df["Fare"]<lower)|
(df["Fare"]>upper)]
print(outliers[["Name","Fare"]])
```

```python
df=df[(df["Fare"]>=lower) & (df["Fare"]<=upper)]
print("numberb of outliers:",len(outliers))
```
